### Analizamos las imagenes

#### Histogramas

Un histograma de imagen es una representación gráfica de la distribución de los niveles de intensidad 
(valores de gris) en una imagen digital. Muestra cuántos píxeles tienen cada valor de intensidad, desde 0 (negro) hasta 255 (blanco).

¿Para qué sirve?
- Permite analizar el contraste de una imagen.
- Ayuda a detectar si la imagen está sobreexpuesta o subexpuesta.
- Es útil para decidir si conviene aplicar filtros o ecualización de histograma.
- En la detección de bordes, un bajo contraste puede dificultar los resultados, por lo que observar el histograma es una herramienta clave.


```python
# Mostrar histogramas de las 3 imágenes originales en escala de grises
gray1 = cv2.cvtColor(imagen1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(imagen2, cv2.COLOR_BGR2GRAY)
gray3 = cv2.cvtColor(imagen3, cv2.COLOR_BGR2GRAY)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
imagenes_grises = [gray1, gray2, gray3]

for ax, img, nombre in zip(axes, imagenes_grises, titulos):
    ax.hist(img.ravel(), bins=256, color='gray')
    ax.set_title(f'Histograma - {nombre}')
    ax.set_xlabel('Intensidad')
    ax.set_ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

## Ecualización de la imagen satelital

### ¿Qué es?

La ecualización del histograma es una técnica de procesamiento de imágenes que mejora el contraste distribuyendo de forma más uniforme los niveles de intensidad.
Consiste en transformar los valores de los píxeles para que el histograma de la imagen resultante sea lo más plano posible, aumentando así el rango dinámico visible.

Es especialmente útil cuando la imagen original tiene un contraste bajo, es decir, sus valores están concentrados en una zona estrecha del histograma. Exactamente es lo que nos pasaba en la imagen satelital de bariloche utilizada.

```python
# Aplicar ecualización del histograma
img_original = gray1
img_ecualizada = cv2.equalizeHist(img_original)

# Mostrar imagen original y ecualizada
fig, axes = plt.subplots(2, 2, figsize=(12, 6))
fig.suptitle('Ecualización de Histograma - Imagen Satelital', fontsize=16)

axes[0, 0].imshow(img_original, cmap='gray')
axes[0, 0].set_title('Imagen Original')
axes[0, 0].axis('off')

axes[0, 1].imshow(img_ecualizada, cmap='gray')
axes[0, 1].set_title('Imagen Ecualizada')
axes[0, 1].axis('off')

axes[1, 0].hist(img_original.ravel(), bins=256, color='gray')
axes[1, 0].set_title('Histograma Original')
axes[1, 0].set_xlabel('Intensidad')
axes[1, 0].set_ylabel('Frecuencia')

axes[1, 1].hist(img_ecualizada.ravel(), bins=256, color='gray')
axes[1, 1].set_title('Histograma Ecualizado')
axes[1, 1].set_xlabel('Intensidad')
axes[1, 1].set_ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

La imagen original presentaba un histograma altamente concentrado en los valores bajos de intensidad, lo cual refleja un contraste limitado: predominaban las zonas oscuras y muy pocas regiones estaban bien iluminadas.

Al aplicar la ecualización del histograma:

- Se redistribuyen los niveles de gris, extendiendo el rango de intensidades.

- El contraste mejora: zonas que antes eran oscuras y homogéneas ahora presentan más detalle.

- El histograma se "aplana", ocupando una mayor porción del eje de intensidad.

Conclusión: la ecualización permitió realzar detalles ocultos en zonas oscuras y mejorar la interpretación visual.

## Aplicamos Prewitt y Sobel

Los vamos a aplicar tanto en las imagenes originales como en sus versiones ecualizadas para observar lo analizado previamente

```python
#Definimos las funciones para aplicar Prewitt y Sobel
def aplicar_prewitt(img):
    kernel_x = np.array([[1, 0, -1],
                         [1, 0, -1],
                         [1, 0, -1]])
    kernel_y = np.array([[1, 1, 1],
                         [0, 0, 0],
                         [-1, -1, -1]])
    gx = cv2.filter2D(img, -1, kernel_x)
    gy = cv2.filter2D(img, -1, kernel_y)
    return np.hypot(gx, gy)

def aplicar_sobel(img):
    gx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
    gy = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
    return np.hypot(gx, gy)
```

```python
# Convertimos la imagen satelital a escala de grises
gray1 = cv2.cvtColor(imagen1, cv2.COLOR_BGR2GRAY)

# Ecualización del histograma
gray1_eq = cv2.equalizeHist(gray1)

# Aplicamos bordes a ambas versiones
prewitt_orig = aplicar_prewitt(gray1)
sobel_orig = aplicar_sobel(gray1)

prewitt_eq = aplicar_prewitt(gray1_eq)
sobel_eq = aplicar_sobel(gray1_eq)

# Mostramos resultados
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Comparación: Imagen Satelital Original vs Ecualizada', fontsize=18)

# Fila 1 - Original
axes[0, 0].imshow(gray1, cmap='gray')
axes[0, 0].set_title('Original (Gris)')
axes[0, 0].axis('off')

axes[0, 1].imshow(prewitt_orig, cmap='gray')
axes[0, 1].set_title('Prewitt - Original')
axes[0, 1].axis('off')

axes[0, 2].imshow(sobel_orig, cmap='gray')
axes[0, 2].set_title('Sobel - Original')
axes[0, 2].axis('off')

# Fila 2 - Ecualizada
axes[1, 0].imshow(gray1_eq, cmap='gray')
axes[1, 0].set_title('Ecualizada')
axes[1, 0].axis('off')

axes[1, 1].imshow(prewitt_eq, cmap='gray')
axes[1, 1].set_title('Prewitt - Ecualizada')
axes[1, 1].axis('off')

axes[1, 2].imshow(sobel_eq, cmap='gray')
axes[1, 2].set_title('Sobel - Ecualizada')
axes[1, 2].axis('off')

plt.tight_layout()
plt.show()
```

## Aplicamos ruido Sal y Pimienta y ruido Gaussiano. Posteriormente volvemos a utilizar los Prewitt y Sobel

Se usan ambos tipos de ruido sobre las imagenes originales en escala de grises para:
- Demostrar diferencias en la sensibilidad de los operadores de gradiente.
- Justificar cuándo conviene aplicar un suavizado previo.

```python
# Definimos la función para agregar ruido sal y pimienta
def agregar_ruido_sal_pimienta(img, prob=0.02): ##agregarle mas prob si queremos que tenga mas imapacto el ruido. Representa la proporción total de píxeles que serán modificados por el ruido.
    noisy = img.copy()
    salt = np.random.rand(*img.shape) < prob / 2
    pepper = np.random.rand(*img.shape) < prob / 2
    noisy[salt] = 255
    noisy[pepper] = 0
    return noisy
```

```python
# Definimos la funcion de ruido gaussiano
def agregar_ruido_gaussiano(img, mean=0, std=20):
    gauss = np.random.normal(mean, std, img.shape).astype(np.float32)
    noisy = img.astype(np.float32) + gauss
    return np.clip(noisy, 0, 255).astype(np.uint8)
```

```python
# Lista de imágenes grises y sus nombres
imagenes = [gray1, gray2, gray3]
nombres = ['Satelital', 'Edificio', 'Rostro']

# Procesar cada imagen
for img, nombre in zip(imagenes, nombres):
    sal_pimienta = agregar_ruido_sal_pimienta(img, prob=0.02)
    gaussiano = agregar_ruido_gaussiano(img, std=20)

    prewitt_sp = aplicar_prewitt(sal_pimienta)
    sobel_sp = aplicar_sobel(sal_pimienta)

    prewitt_gauss = aplicar_prewitt(gaussiano)
    sobel_gauss = aplicar_sobel(gaussiano)

    # Mostrar resultados 
    fig, axes = plt.subplots(3, 3, figsize=(18, 12))
    fig.suptitle(f'Análisis con ruido - {nombre}', fontsize=18)

    axes[0, 0].imshow(img, cmap='gray')
    axes[0, 0].set_title("Original", fontsize=14)
    axes[0, 0].axis('off')

    axes[0, 1].imshow(prewitt_gauss, cmap='gray')
    axes[0, 1].set_title("Prewitt", fontsize=14)
    axes[0, 1].axis('off')

    axes[0, 2].imshow(sobel_gauss, cmap='gray')
    axes[0, 2].set_title("Sobel", fontsize=14)
    axes[0, 2].axis('off')

    axes[1, 0].imshow(gaussiano, cmap='gray')
    axes[1, 0].set_title("Ruido Gaussiano", fontsize=14)
    axes[1, 0].axis('off')

    axes[1, 1].imshow(prewitt_sp, cmap='gray')
    axes[1, 1].set_title("Prewitt + Gauss", fontsize=14)
    axes[1, 1].axis('off')

    axes[1, 2].imshow(sobel_sp, cmap='gray')
    axes[1, 2].set_title("Sobel + Gauss", fontsize=14)
    axes[1, 2].axis('off')

    axes[2, 0].imshow(sal_pimienta, cmap='gray')
    axes[2, 0].set_title("Ruido Sal y Pimienta", fontsize=14)
    axes[2, 0].axis('off')
    
    axes[2, 1].imshow(prewitt_sp, cmap='gray')
    axes[2, 1].set_title("Prewitt + S&P", fontsize=14)
    axes[2, 1].axis('off')

    axes[2, 2].imshow(sobel_sp, cmap='gray')
    axes[2, 2].set_title("Sobel + S&P", fontsize=14)
    axes[2, 2].axis('off')

    for ax in axes[2, 1:]:
        ax.axis('off')

    plt.tight_layout()
    plt.show()
```

### Análisis de Bordes con Ruido (Prewitt y Sobel)

#### Imagen Satelital

  - **Sobel** más robusto: mantiene mejor definición.

  - **Sobel** ligeramente más tolerante al ruido.

#### Imagen de Edificio

  - **Sobel** resiste mejor, pero la calidad general disminuye.

---

**Conclusión general:**

## Aplicación de filtro Gaussiano previo a los operadores de borde


La suavización con un filtro Gaussiano ayuda a **reducir el ruido** en la imagen. Cuando se aplica un operador de borde como Sobel o Prewitt directamente sobre una imagen ruidosa, se pueden detectar **falsos bordes** (bordes que en realidad no existen en la escena original). Al suavizar:

- Se atenúan variaciones pequeñas (como el ruido),
- Se preservan los contornos grandes y reales.

Esto **mejora la robustez del operador de borde** y da lugar a una detección más coherente.

```python
# Aplicar filtro Gaussiano y luego Prewitt y Sobel
imagenes_grises = [gray1, gray2, gray3]
nombres = ['Satelital', 'Edificio', 'Rostro']

for i, (img, nombre) in enumerate(zip(imagenes_grises, nombres)):
    img_gauss = cv2.GaussianBlur(img, (5, 5), sigmaX=1)
    prewitt = aplicar_prewitt(img_gauss)
    sobel = aplicar_sobel(img_gauss)

    # Mostrar resultados
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))
    fig.suptitle(f'Filtro Gaussiano + Bordes - {nombre}', fontsize=16)

    axes[0].imshow(img_gauss, cmap='gray')
    axes[0].set_title('Imagen Suavizada (Gauss)')
    axes[0].axis('off')

    axes[1].imshow(prewitt, cmap='gray')
    axes[1].set_title('Bordes con Prewitt')
    axes[1].axis('off')

    axes[2].imshow(sobel, cmap='gray')
    axes[2].set_title('Bordes con Sobel')
    axes[2].axis('off')

    plt.tight_layout()
    plt.show()
```

## Conteo de píxeles detectados como bordes

Además del análisis visual de los bordes detectados, se implementó una métrica objetiva para cuantificar la cantidad de bordes detectados por cada operador.

Para ello:

- Se aplicó un filtro Gaussiano como paso previo, con el objetivo de reducir el ruido y facilitar una detección más robusta.

- Luego se aplicaron los operadores Prewitt y Sobel para detectar bordes.

- Finalmente, se utilizó un proceso de binarización con un umbral fijo (50) para separar los píxeles de borde (intensidad alta) del resto.

- Se contabilizaron los píxeles que superan el umbral, interpretándolos como cantidad de bordes detectados.

Este conteo permite:

- Comparar cuantitativamente el desempeño de cada operador.

- Analizar cómo varía la detección según la imagen (satelital, rostro, edificio).

- Estimar la sensibilidad del operador a distintas estructuras y contrastes.

El valor del umbral (50) lo elegí de forma empírica para este conjunto de imágenes. Si bien podría optimizarse, se mantuvo constante para poder comparar los resultados entre imágenes de manera homogénea.

```python
# Función para binarizar y contar píxeles de borde
def binarizar_y_contar(img, umbral=50):
    binaria = (img > umbral).astype(np.uint8)
    return binaria * 255, np.sum(binaria)

# Aplicar binarización y conteo sobre cada imagen suavizada
for i, (img, nombre) in enumerate(zip(imagenes_grises, nombres)):
    img_gauss = cv2.GaussianBlur(img, (5, 5), sigmaX=1)

    prewitt = aplicar_prewitt(img_gauss)
    sobel = aplicar_sobel(img_gauss)

    bin_pre, count_pre = binarizar_y_contar(prewitt)
    bin_sob, count_sob = binarizar_y_contar(sobel)

    # Mostrar resultados
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))
    fig.suptitle(f'Conteo de Bordes - {nombre}', fontsize=16)

    axes[0].imshow(img_gauss, cmap='gray')
    axes[0].set_title('Imagen Suavizada')
    axes[0].axis('off')

    axes[1].imshow(bin_pre, cmap='gray')
    axes[1].set_title(f'Prewitt - Bordes: {count_pre}')
    axes[1].axis('off')

    axes[2].imshow(bin_sob, cmap='gray')
    axes[2].set_title(f'Sobel - Bordes: {count_sob}')
    axes[2].axis('off')

    plt.tight_layout()
    plt.show()
```

### Conclusiones generales

A lo largo del trabajo se exploraron distintas técnicas de detección de bordes aplicadas a imágenes reales con distintas características (satelital, edificio y rostro). Se analizaron los resultados de los operadores de gradiente de Prewitt y Sobel bajo distintas condiciones: imágenes originales, suavizadas con filtro Gaussiano, ecualizadas y contaminadas con ruidos (sal y pimienta, gaussiano).

Se observó que:

- La ecualización del histograma permitió mejorar el contraste en imágenes con bajo rango dinámico, facilitando la detección. En nuestro caso fueron la imagen satelital y la del edificio Guadi.

- El suavizado previo mejora la detección de bordes, reduciendo falsos positivos por ruido.

- El ruido afecta negativamente el resultado, pero Sobel mostró mayor robustez que Prewitt.

- El conteo de bordes permitió una cuantificación objetiva del impacto de cada técnica.

En resumen, combinar técnicas como ecualización, suavizado y análisis por gradiente permite mejorar significativamente la detección de bordes en contextos diversos, lo que resulta clave para aplicaciones como segmentación, clasificación o análisis automático de imágenes.
